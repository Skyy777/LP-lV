```python
def grocery_chatbot():
    print("🤖 Welcome to FreshMart Grocery Store!")
    print("Type 'menu' to see items or 'exit' to quit.\n")
    products = {
        "fruits": {"apple": 50, "banana": 30, "orange": 40},
        "vegetables": {"potato": 20, "tomato": 25, "onion": 30},
        "dairy": {"milk": 60, "cheese": 120, "butter": 80}
    }
    cart = {}
    while True:
        user_input = input("You: ").lower()
        if user_input == "exit":
            print("🤖 Thank you for visiting FreshMart! 🛍️")
            break
        elif user_input == "menu":
            print("\n📋 Available Categories:")
            for category in products:
                print(f"- {category}")
            print("Type a category name to see items.\n")
        elif user_input in products:
            print(f"\n🛍️ {user_input.title()} Items:")
            for item, price in products[user_input].items():
                print(f"{item} - ₹{price}")
            print("Type item name to add to cart.\n")
        elif any(user_input in items for items in products.values()):
            for category in products:
                if user_input in products[category]:
                    cart[user_input] = cart.get(user_input, 0) + 1
                    print(f"✅ {user_input} added to cart.\n")
        elif user_input == "cart":
            print("\n🧺 Your Cart:")
            total = 0
            for item, qty in cart.items():
                for category in products:
                    if item in products[category]:
                        price = products[category][item]
                        total += price * qty
                        print(f"{item} x{qty} = ₹{price * qty}")
            print(f"💰 Total: ₹{total}\n")
        else:
            print("🤖 Sorry, I didn’t understand. Try 'menu', 'cart', or 'exit'.\n")
grocery_chatbot()
```
