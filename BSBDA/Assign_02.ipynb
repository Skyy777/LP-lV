```python
import numpy as np
Data1 = pd.read_csv('AcademicPerformance.csv')
print(Data1)
```

```python
import statistics
Data1 = pd.read_csv('AcademicPerformance.csv')
print(Data1)
```

```python
Data1['raisedhands'] = Data1['raisedhands'].replace(np.NaN,statistics.mode(Data1['raisedhands']))
print(Data1['raisedhands'][:10])
```
