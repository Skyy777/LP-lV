# Data Wrangling-I using Python

Iris Dataset Analysis

```python
# Import Libraries
import pandas as pd
import numpy as np
```

```python
# Load Dataset
df = pd.read_csv('iris.csv')
df.head()
```

```python
# Dataset Info
df.info()
```

```python
# Shape of dataset
df.shape
```

```python
# Missing Values
df.isnull().sum()
```

```python
# Statistical Summary
df.describe()
```

```python
# Data Types
df.dtypes
```

```python
# Convert species to category
df['species'] = df['species'].astype('category')
df.dtypes
```

```python
# Normalize numerical columns
numeric_cols = df.select_dtypes(include=['float64', 'int64']).columns

df[numeric_cols] = (df[numeric_cols] - df[numeric_cols].min()) / (df[numeric_cols].max() - df[numeric_cols].min())

df.head()
```

```python
# One-hot encoding
df_encoded = pd.get_dummies(df, columns=['species'])
df_encoded.head()
```

```python
# NumPy Calculations
print("Mean:\n", np.mean(df[numeric_cols]))
print("\nStd Dev:\n", np.std(df[numeric_cols]))
```
